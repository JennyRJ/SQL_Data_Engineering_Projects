```python
#install the datasets library(iff not yet installed: - !pip list to check)
!pip install datasets

#import libraries required
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

#load datasets

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset["train"].to_pandas()

#Data cleaning
df["job_posted_date"] =pd.to_datetime(df["job_posted_date"])
```

```python
df.describe()
```

```python
df.head(10)
```

```python
date_counts = df.job_posted_date.value_counts()
```

```python
date_counts
```

```python
date_counts =date_counts.sort_index()
```

```python
plt.plot(date_counts.index, date_counts)
```

```python
df["job_posted_month"] = df["job_posted_date"].dt.month
```

```python
df.head(5)
```

```python
monthly_counts = df.job_posted_month.value_counts()
monthly_counts =monthly_counts.sort_index()
monthly_counts
```

```python
plt.plot(monthly_counts.index, monthly_counts)
```

Visualize salaries median salaries based on job title:Highpaying and low paying roles

```python
job_salary = df.groupby(["job_title_short"])["salary_year_avg"].median().sort_values()
```

```python
job_salary
```

```python
job_salary.plot(kind='barh')
plt.title("Median salary by job title")
plt.xticks(rotation=45,ha='right')
plt.xlabel("salary(USD)")
plt.ylabel("")
plt.show()
```
